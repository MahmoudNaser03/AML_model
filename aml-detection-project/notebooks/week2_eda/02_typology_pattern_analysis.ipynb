```python
import pandas as pd
 
columns = ['Timestamp','From Bank','Account','To Bank','Account.1',
           'Amount Received','Receiving Currency','Amount Paid',
           'Payment Currency','Payment Format','Is Laundering']
 
records = []
current_typology = None
current_pattern_id = 0
 
with open("../data/raw/HI-Small_Patterns.txt") as f:
    for line in f:
        line = line.strip()
        if line.startswith("BEGIN LAUNDERING ATTEMPT"):
            current_typology = line.split(" - ")[1].split(":")[0].strip()
            current_pattern_id += 1
        elif line.startswith("END LAUNDERING ATTEMPT"):
            current_typology = None
        elif line and current_typology:
            parts = line.split(",")
            if len(parts) == 11:
                row = dict(zip(columns, parts))
                row['Typology'] = current_typology
                row['Pattern_ID'] = current_pattern_id
                records.append(row)
 
patterns_df = pd.DataFrame(records)
print(patterns_df.shape)
patterns_df['Typology'].value_counts()
```

### One example pattern (all its transactions) per typology

```python
for typ in patterns_df['Typology'].unique():
    example_id = patterns_df[patterns_df['Typology'] == typ]['Pattern_ID'].iloc[0]
    print(f"\n=== {typ} — Pattern #{example_id} ===")
    display(patterns_df[patterns_df['Pattern_ID'] == example_id][
        ['Timestamp','Account','Account.1','Amount Paid','Payment Currency','Payment Format']
    ])
```
